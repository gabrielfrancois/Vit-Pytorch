```python
import torch
import platform
import psutil

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device =", device)

if device.type == "cuda":
    print("Nom du GPU :", torch.cuda.get_device_name(0))
    props = torch.cuda.get_device_properties(0)
    print("Total GPU Memory (GB):", props.total_memory / 1024 ** 3)
    print("Multi Processor Count :", props.multi_processor_count)
    print("CUDA Capability      :", f"{props.major}.{props.minor}")
else:
    print("Vous êtes sur le CPU")
    print("Nom du CPU :", platform.processor())
    print("Nb de cœurs (physiques) :", psutil.cpu_count(logical=False))
    print("Nb de cœurs (total/logiques):", psutil.cpu_count(logical=True))
    print("RAM totale (GB) :", psutil.virtual_memory().total / 1024 ** 3)
```
